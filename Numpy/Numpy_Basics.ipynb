#### 🔷 **Introduction to NumPy: What is NumPy?**

NumPy is a Python library used to work with numbers, arrays, and mathematical data in a fast and efficient way.

It provides a special data structure called an array, which is much more powerful than Python lists.

In simple words:

NumPy helps us handle large numerical data quickly and easily.

#### **Why Do We Use NumPy?**

When we work with data, machine learning, or scientific problems, we usually deal with:

* Thousands or millions of numbers

* Tables (rows & columns)

* Images (pixels)

* Mathematical formulas

Python lists are not designed for this kind of heavy numerical work.
NumPy is specially built for performance and accuracy.

#### **Purpose of Using NumPy**

We use NumPy to:

* Store data in 1D, 2D, and multi-dimensional arrays

* Perform fast mathematical operations

* Work with vectors, matrices, and tensors

* Reduce code complexity

* Improve performance and memory usage

#### **History of NumPy (Numerical Python)**

##### **When was NumPy created?**


* NumPy was officially released in **2006**.


#### **Who created NumPy?**


* NumPy was created by **Travis Oliphant**, a mathematician and software developer.


#### **Why did Travis Oliphant think of creating NumPy?**


* Before NumPy existed, Python was slow for numerical and scientific computations.


At that time:


* Python did not have a powerful array system

* Scientists and engineers were using MATLAB, Fortran, and C for fast math

* Python was easy to write, but too slow for large numerical data


There were two older libraries:


* Numeric

* Numarray


#### **Both had problems:**


* Different APIs (confusing)

* Performance limitations

* Not well-maintained together


👉 Travis Oliphant noticed this problem and thought:


**“Why not create ONE powerful, fast, and unified numerical library for Python?”**


That idea became NumPy.

## **NumPy Foundations & Array Basics**

### Topics Covered:
1. Creating arrays from lists
2. List vs NumPy array comparison
3. Creating arrays from scratch (zeros, ones, full, random, arange)
4. Vector, Matrix, and Tensor
5. Array properties (shape, ndim, size, dtype)

```python
import numpy as np
```

### **Creating arrays from lists**

```python
arr_1d = np.array([1, 2, 3, 4, 5])
print(f"1 Dimensional array: {arr_1d}")

arr_2d = np.array([[1, 2, 3],
                   [4, 5, 6]])
print(f"2Dimensional array: {arr_2d}")

arr_3d = np.array([[[1, 2, 4], [3, 4, 5],
                    [5, 6, 9], [7, 8, 10]]])

print(f"3Dimensional array: {arr_3d}")

print(arr_1d.shape)
print(arr_2d.shape)
print(arr_3d.shape)
```

### **List vs NumPy array comparison**

```python
py_list = [1, 2, 3, 4, 5]
print(f"Multiplication: {py_list * 2}")

np_array = np.array([1, 2, 3, 4, 5])
print(f"Multiplication array: {np_array * 2}")

import time
start_time = time.time()

py_list = []

for i in range(1_000_000):
    value = i * 2
    py_list.append(value)
py_time = time.time() - start_time

print(f"Python list time: {py_time} in seconds")

start_time = time.time()

np_array = np.arange(1_000_000)

np_time = time.time() - start_time
print(f"Numpy array time: {np_time} in seconds")
```

### **Creating arrays from scratch (zeros, ones, full, random, arange)**

```python
zeros = np.zeros((2, 3))   # (rows, columns)
print(f"Zeros array: {zeros}")

# int("9")
ones = np.ones((4, 5), dtype='int64')
print(f"Ones array: {ones}")

full = np.full((4, 5), 99)
print(f"Full array: {full}")


random = np.random.random((3, 4))
print(f"\nRandom array: {random}")

sequence = np.arange(1, 12, 2)  # (start, stop, step)
print(f"\nSequence array: {sequence}")
```

### **Vector, Matrix, and Tensor**

```python
arr_1d = np.array([1, 2, 3, 4, 5])
print(f"1 Dimensional array: {arr_1d}")

arr_2d = np.array([[1, 2, 3],
                   [4, 5, 6]])
print(f"2Dimensional array: {arr_2d}")

arr_3d = np.array([[[1, 2, 4], [3, 4, 5],
                    [5, 6, 9], [7, 8, 10]]])

print(f"3Dimensional array: {arr_3d}")
```

### **Array properties (shape, ndim, size, dtype)**

```python
arr_2d = np.array([[1, 2, 3],
                   [4, 5, 6]])


print(f"Shape of array: {arr_2d.shape}")
print(f"Shape of array: {arr_2d.ndim}")
print(f"Shape of array: {arr_2d.size}")
print(f"Shape of array: {arr_2d.dtype}")
```
